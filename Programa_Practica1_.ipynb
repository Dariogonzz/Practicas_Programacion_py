```python
# Programa_Practica1_9/10/26
# Dario González Moreno 51151307
# Héctor Aguirre Salinas


# 1. Creamos el almacén (diccionario) antes del ciclo
base_de_datos = {}
continuar = True

print("Bienvenido al SISTEMA AVANZADO FASTOBOX")

while continuar == True:
    print("\n-----------------------------------")
    print("Opcion 1: Registrar Paquete")
    print("Opcion 2: Despachar Paquete")
    print("Opcion 3: Ver Estado Detallado De Las Rutas")
    print("Opcion 4: Salir")

    # Leemos la opción como texto
    opcion_usuario = input("\nIngrese la opcion que desee: ")

    if opcion_usuario == "1":
        print("\n--- REGISTRAR PAQUETE ---")
        codigo = input("Ingrese el código del paquete (ej. P01): ")

        if codigo in base_de_datos:
            print("Error: Ya existe un paquete con ese código.")
        else:
            destinatario = input("Nombre del destinatario: ")
            destino = input("Ciudad de destino: ")

            # Guardamos: [Destinatario, Destino, Estado]
            base_de_datos[codigo] = [destinatario, destino, "En almacén"]
            print("¡Paquete registrado correctamente!")

    elif opcion_usuario == "2":
        print("\n--- DESPACHAR PAQUETE ---")
        codigo = input("Ingrese el código del paquete a despachar: ")

        # Validamos si el paquete existe antes de intentar modificarlo
        if codigo in base_de_datos:
            # Comprobamos si ya fue despachado antes
            if base_de_datos[codigo][2] == "Despachado":
                print("El paquete ya había sido despachado anteriormente.")
            else:
                # Modificamos el estado (posición 2 de la lista)
                base_de_datos[codigo][2] = "Despachado"
                print(f"¡Éxito! El paquete {codigo} ha salido hacia su destino.")
        else:
            print("Error: No se encontró ningún paquete con ese código.")

    elif opcion_usuario == "3":
        print("\n--- ESTADO DETALLADO DE LOS PAQUETES ---")

        # Comprobamos si el diccionario está vacío
        if len(base_de_datos) == 0:
            print("Aún no hay ningún paquete registrado en el sistema.")
        else:
            # Recorremos el diccionario para imprimir paquete por paquete
            for codigo in base_de_datos:
                datos = base_de_datos[codigo]
                destinatario = datos[0]
                destino = datos[1]
                estado = datos[2]

                print(f"ID: {codigo} | Destinatario: {destinatario} | Destino: {destino} | Estado: {estado}")

    elif opcion_usuario == "4":
        print("Saliendo del sistema...")
        continuar = False

    else:
        print("Error: Opción no válida. Ingresa 1, 2, 3 o 4.")
```
